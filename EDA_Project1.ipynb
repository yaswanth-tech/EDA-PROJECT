```python
# Viewing and Filtering

# 1. Select and display only Age column
print(df["Age"])
print("_________________________")

# 2. Select Age, Education, Occupation and Income columns
print(df[["Age", "Education", "Occupation", "Income"]])
print("_________________________")

# 3. Display first 10 rows
print(df.iloc[:10])
print("_________________________")

# 4. Select rows 10 to 20 and display Age and Education
print(df.iloc[10:21][["Age", "Education"]])
print("_________________________")

# 5. People whose age is greater than 50
print(df[df["Age"] > 50])
print("_________________________")

# 6. People whose income is >50K
print(df[df["Income"].str.strip() == ">50K"])
print("_________________________")

# 7. Age > 40 and Hours per Week > 40
print(df[(df["Age"] > 40) & (df["Hours per Week"] > 40)])
print("_________________________")

# 8. Education is Bachelors or Masters
print(df[df["Education"].str.strip().isin(["Bachelors", "Masters"])])
print("_________________________")

# 9. Mean, Median, Minimum and Maximum Age
print("Mean:", df["Age"].mean())
print("Median:", df["Age"].median())
print("Minimum:", df["Age"].min())
print("Maximum:", df["Age"].max())
print("_________________________")

# 10. Standard deviation of Age and Hours per Week
print("Age Std:", df["Age"].std())
print("Hours per Week Std:", df["Hours per Week"].std())
print("_________________________")

# 11. Statistical summary of numerical columns
print(df.describe())
print("_________________________")

# 12. Unique values and frequency in Education
print(df["Education"].value_counts())
print("_________________________")

# 13. Number of people in each Income category
print(df["Income"].value_counts())
print("_________________________")

# 14. Percentage of people in each Income category
print(df["Income"].value_counts(normalize=True) * 100)
print("_________________________")

# 15. Average Hours per Week for income >50K
print(df[df["Income"].str.strip() == ">50K"]["Hours per Week"].mean())
print("_________________________")

# 16. Average Age of people with Bachelors
print(df[df["Education"].str.strip() == "Bachelors"]["Age"].mean())
print("_________________________")

# 17. Youngest and oldest person
print("Youngest:", df["Age"].min())
print("Oldest:", df["Age"].max())
print("_________________________")

# 18. Correlation between numerical columns
print(df.corr(numeric_only=True))
print("_________________________")

# 19. Most common Education and Occupation
print("Education Mode:", df["Education"].mode()[0])
print("Occupation Mode:", df["Occupation"].mode()[0])
```

```python
import pandas as pd

df = pd.read_csv("adult.csv")

# Cleaning

# Approach-1: Delete the records (rows)
df = df.dropna()
print(df)
print("_________________________")

# Approach-2: Delete the columns
df = df.dropna(axis=1)
print(df)
print("_________________________")
```

```python
df['Age']=df['Age'].bfill()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].bfill()
print(df)
df['Ocuupation'] = df['Occupation'].bfill()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].bfill()
print(df)
```

```python
df['Age']=df['Age'].ffill()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].ffill()
print(df)
df['Ocuupation'] = df['Occupation'].ffill()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].ffill()
print(df)
```

```python
df['Age']=df['Age'].interpolate()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].interpolate()
print(df)
df['Ocuupation'] = df['Occupation'].interpolate()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].interpolate()
print(df)
```

```python
#dectecting outliers bounds(lower & upper)
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3-Q1

print('Q1',Q1)
print('Q3',Q3)
print('IQR',IQR)

lower_bound = (Q1-1.5*IQR)
upper_bound = (Q3+1.5*IQR)
print(lower_bound)
print(upper_bound)
print("_________________________")
#identify IQR Outliers
iqr_outliers = df[
    (df['Age']<lower_bound) |
    (df['Age']>upper_bound)
]

print('IQR OUTLIERS:',len(iqr_outliers))
```

```python
#z-score method
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
df['Age_zscore'] = zscore(df['Age'])

z_outliers = df[
    df['Age_zscore'].abs() > 3
]

print("Z-Score Outliers: ",len(z_outliers))
```

```python
#z-score-trimmed
df_trimmed = df[df['Age_zscore'].abs()<=3]
print(df_trimmed)
```

```python
#winsorization
df['Age_winsorized']=winsorize(
    df['Age'],
    limits=[0.05,0.5]
)
print(df.head())
```
